```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# 1. Generate a range of dates (e.g., from year 2000 to 2040)
dates = pd.date_range(start="1900-01-01", end="2100-12-31", freq="M")

df = pd.DataFrame()
df['date_key'] = dates.date
df['year_int'] = dates.year
df['month_num'] = dates.month
df['month_name'] = dates.strftime('%B')
df['year_month_key'] = dates.year * 100 + dates.month
df['quarter_num'] = dates.quarter
df['quarter_name'] = 'Q' + dates.quarter.astype(str)
df['year_quarter_key'] = dates.year.astype(str) + '-Q' + dates.quarter.astype(str)

# 2. Push to your PostgreSQL database
# engine = create_engine('postgresql://username:password@localhost:5432/your_db')
# df.to_sql('dim_date', engine, if_exists='append', index=False)
```
