```python
import requests
import os
import keyring, getpass
```

```python
service = "api.alpaca.markets"
api_key = keyring.get_password(service, "ALPACA_API_KEY")
api_secret = keyring.get_password(service, "ALPACA_SECRET_KEY")

print(f"API Key: {api_key[:3]}")
print(f"API Secret: {api_secret[:3]}")
```

```python
def get_alpaca_asset_list(status="active", asset_class="us_equity", api_key="", api_secret=""):
    base_url = "https://api.alpaca.markets/v2/assets"
    
    headers = {
        "accept": "application/json",
        "APCA-API-KEY-ID": api_key,
        "APCA-API-SECRET-KEY": api_secret,
    }

    params = {
        "status": status,
        "asset_class": asset_class,
    }

    response = requests.get(base_url, headers=headers, params=params)
    return response

    
```

```python
asset_classes = ["us_equity", "crypto"]
```

```python
active_crypto_assets = get_alpaca_asset_list(status="active", 
                                            asset_class="crypto", 
                                            api_key=api_key, 
                                            api_secret=api_secret)
active_us_equity_assets = get_alpaca_asset_list(status="active", 
                                            asset_class="us_equity", 
                                            api_key=api_key, 
                                            api_secret=api_secret)

for item in active_crypto_assets.json():
    print(item)
```
