```python
service = "api.alpaca.markets"
API_KEY = keyring.get_password(service, "ALPACA_API_KEY")
SECRET_KEY = keyring.get_password(service, "ALPACA_SECRET_KEY")

print(f"API Key: {API_KEY[:3]}")
print(f"API Secret: {SECRET_KEY[:3]}")
```

```python
#import pandas as pd
from alpaca.data.historical.screener import ScreenerClient
from alpaca.data.requests import MostActivesRequest
from alpaca.data.enums import MostActivesBy

# 1. Initialize the Screener Client
# Replace with your actual Alpaca API keys or environment variables
#API_KEY = api_key
#SECRET_KEY = api_secret

screener_client = ScreenerClient(api_key=API_KEY, secret_key=SECRET_KEY)

# 2. Configure your liquidity query
# You can sort 'by' MostActivesBy.VOLUME or MostActivesBy.TRADES
request_params = MostActivesRequest(
    by=MostActivesBy.VOLUME,
    top=100  # Supports fetching between 1 and 100 assets
)

# 3. Fetch data from Alpaca
try:
    most_active_data = screener_client.get_most_actives(request_params)
    
    # 4. Parse the results into a Pandas DataFrame for easier viewing
    liquidity_list = []
    for asset in most_active_data.most_actives:
        liquidity_list.append({
            "Symbol": asset.symbol,
            "Daily Volume": asset.volume,
            "Trade Count": asset.trade_count
        })
        
    df = pd.DataFrame(liquidity_list)
    print(f"Top Liquidity Assets (Last Updated: {most_active_data.last_updated})")
    print(df.to_string(index=False))

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
from io import StringIO

import pandas as pd
import requests

url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

headers = {
    "User-Agent": "FinancialTickerUpdater/1.0 (contact: your-email@example.com)"
}

response = requests.get(url, headers=headers, timeout=30)
response.raise_for_status()

tables = pd.read_html(StringIO(response.text))
sp500_df = tables[0]

tickers = sp500_df["Symbol"].tolist()
```

```python
tickers
```

```python
#let's write the tickers list to a txt in documentation
#with open("../../documentation/sp500_tickers.txt", "w") as f:
#    for ticker in tickers:
#        f.write(f"{ticker}\n")
old_tickers = []
with open("../../documentation/sp500_tickers.txt", "r") as f:
    old_tickers = [line.strip() for line in f.readlines()]

old_tickers
```

```python
today = datetime.today()
from time import sleep
```

```python
for ticker in tickers:
    try:
        request_params = StockBarsRequest(
            symbol_or_symbols=[ticker],
            timeframe=TimeFrame.Day,
            start=datetime(1990, 1, 1),
            end=today
        )

        # Fetch data and convert to a pandas DataFrame
        bars = client.get_stock_bars(request_params)
        df = bars.df.reset_index().drop(columns=['symbol'])
        db.set_table_name(ticker.lower())
        db.set_pdf(df)
        db.write_pdf_to_sql()
        sleep(5)
    except Exception as e:
        print(f"Error processing ticker {ticker}: {e}")
```

```python
engine = db.get_engine()
from sqlalchemy import text
from datetime import timedelta, datetime
```

```python
new_tickers = []
with engine.begin() as connection:
    for ticker in tickers:
        if ticker not in old_tickers:
            new_tickers.append(ticker)
            start_date = datetime(1990, 1, 1)
            request_params = StockBarsRequest(
                symbol_or_symbols=[ticker],
                timeframe=TimeFrame.Day,
                start=start_date,
                end=today
            )
            bars = client.get_stock_bars(request_params)
            df_to_append = bars.df.reset_index().drop(columns=['symbol'])
            if len(df_to_append) > 0:
                db.set_table_name(ticker.lower())
                db.set_pdf(df_to_append)
                db.write_pdf_to_sql(if_exists='append')
                sleep(1)
        else:
            max_date_query = text(f'SELECT MAX(timestamp) FROM "{ticker.lower()}"')
            result = connection.execute(max_date_query)
            max_date = result.scalar()
            print(f"Max date for {ticker}: {max_date}")
            #print(type((max_date + timedelta(days=1)).date()), type(today.date()))
            print('new range:', max_date + timedelta(days=1), today)
            if today.date() > (max_date + timedelta(days=1)).date():
                #print(f"Data for {ticker} is outdated. Needs update.")
                request_params = StockBarsRequest(
                    symbol_or_symbols=[ticker],
                    timeframe=TimeFrame.Day,
                    start=max_date + timedelta(days=1),
                    end=today
                )
                bars = client.get_stock_bars(request_params)
                df_to_append = bars.df.reset_index().drop(columns=['symbol'])
                df_to_append = df_to_append[df_to_append['timestamp'] > max_date]
                if len(df_to_append) > 0:
                    db.set_table_name(ticker.lower())
                    db.set_pdf(df_to_append)
                    db.write_pdf_to_sql(if_exists='append')
                    sleep(1)

with open("../../documentation/sp500_tickers.txt", "a") as f:
    for ticker in new_tickers:
        f.write(f"{ticker}\n")
```
