```python
def convert_series_to_df(series):
    df = series.to_frame().reset_index()
    df.columns = ['date', 'value']
    return df
```

```python
test_frames = {}
for id, series in current_data.items():
    test_frames[id] = convert_series_to_df(series)
```

```python
#now let's use the pgsql class from pgsql.py to write these dataframes to our sql db
from db.pgsql import pgsql
from sqlalchemy import text
```

```python
db = pgsql()
engine = db.get_engine()

with engine.begin() as connection:
    for id in (fred_ids + ['CBBCHUSD','CBBTCUSD', 'CBETHUSD', 'CBLTCUSD']):
        query = text(f'DROP TABLE IF EXISTS "{id}"')
        connection.execute(query)
        query = text(f'DROP TABLE IF EXISTS "{id.lower()}"')
        connection.execute(query)

for id, frame in test_frames.items():
    db.set_table_name(id.lower())
    db.set_pdf(frame)
    db.write_pdf_to_sql()
    
    
```

```python
date_delta_dict = {'D': pd.Timedelta(days=1),
                   'W': pd.Timedelta(weeks=1),
                   'M': pd.Timedelta(days=30),
                   'Q': pd.Timedelta(days=91)}
```

```python
today > date_delta_dict['D'] + pd.Timestamp(max_date)
```

```python
with engine.begin() as connection:
    for id in fred_ids:
        max_date_query = text(f"SELECT MAX(date) FROM {id.lower()}")
        result = connection.execute(max_date_query)
        max_date = pd.Timestamp(result.scalar())
        print(f"Max date for {id}: {max_date}")
        frequency_short_query = text(f"SELECT frequency_short FROM fred_series_info WHERE id = '{id}' LIMIT 1")
        frequency_short_result = connection.execute(frequency_short_query)
        frequency_short = frequency_short_result.scalar()
        print(f"Frequency short for {id}: {frequency_short}")
        next_read_day = max_date + date_delta_dict[frequency_short]
        print(f"Next read day for {id}: {next_read_day}")
        if next_read_day <= today:
            print(f"Fetching data for {id} starting from {next_read_day}")
            #next_start_date = next_read_day
            #max_date + pd.Timedelta(days=1)
            temp_fresh_data = fred.get_series(id, 
                                              observation_start=next_read_day.strftime('%Y-%m-%d'),
                                              observation_end=today_formatted)
            print(f"--- {id} Series ---")
            if temp_fresh_data.empty:
                print(f"No new data returned for {id}, skipping")
                continue
            db.set_table_name(id.lower())
            df_to_append = convert_series_to_df(temp_fresh_data)
            df_to_append = df_to_append[df_to_append['date'] > pd.Timestamp(max_date)]
            #df_to_append['date'] = pd.to_datetime(df_to_append['date'])
            if len(df_to_append) == 0:
                print(f"No new data to append for {id}, skipping")
                continue
            db.set_pdf(df_to_append)
            db.write_pdf_to_sql(if_exists='append')
            sleep(1)
```
